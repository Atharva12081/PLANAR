4. Cluster and validate (silhouette, stability ARI, eta^2 bias audits).
5. Simulate transits and evaluate ROC/AUC.

**FITS cube handling note**
- Use `data[0]` (or a controlled `squeeze`) to retain a `(600, 600)` image.
- Avoid averaging over spatial axes; that collapses morphology.
- You can inspect the cube with `astropy.io.fits.info(path)` before extraction.
